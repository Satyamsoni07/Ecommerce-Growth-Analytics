```python
#Compare ratings across delay groups

delay_rating_summary = (late_reviews.groupby("delay_group", observed=False)
    .agg(orders=("order_id", "count"),average_rating=("review_score", "mean"),low_rating_rate=("is_low_rating", "mean")
         ))

delay_rating_summary["low_rating_rate"] *= 100

delay_rating_summary.round(2)
```

```python
# In this True means the review was created on an earlier calendar day than delivery, which is much more defensible.

delivered_reviews["review_before_delivery"] = (delivered_reviews["review_creation_date"].dt.date
                                               <delivered_reviews["order_delivered_customer_date"].dt.date)

print("Reviews created before delivery:",delivered_reviews["review_before_delivery"].sum())

print("Percentage:",round(delivered_reviews["review_before_delivery"].mean() * 100,2),"%")
```

```python
#Interpret the geolocation grain

geo_location_consistency = (geolocation.groupby("geolocation_zip_code_prefix")
    .agg(cities=("geolocation_city", "nunique"),states=("geolocation_state", "nunique")))

print("ZIP prefixes linked to multiple cities:",(geo_location_consistency["cities"] > 1).sum())

print("ZIP prefixes linked to multiple states:",(geo_location_consistency["states"] > 1).sum())
```

```python
#ZIP-location consistency

multi_state_zips = geo_location_consistency[geo_location_consistency["states"] > 1].index

geolocation[geolocation["geolocation_zip_code_prefix"].isin(multi_state_zips)][[
        "geolocation_zip_code_prefix","geolocation_city","geolocation_state"]].drop_duplicates().sort_values(
    "geolocation_zip_code_prefix")
```

```python
#Creating one row per ZIP prefix

geo_summary = (geolocation.groupby("geolocation_zip_code_prefix", as_index=False)
    .agg(latitude=("geolocation_lat", "median"),longitude=("geolocation_lng", "median")))

geo_summary.head()
```

```python
# Adding the most common city and state

geo_city_state = (geolocation.groupby("geolocation_zip_code_prefix")
    .agg(city=("geolocation_city", lambda x: x.mode().iloc[0]),state=("geolocation_state", lambda x: x.mode().iloc[0]))
    .reset_index())

geo_city_state.head()
```

```python
# Combine  into one clean geolocation table

geo_clean = geo_summary.merge(geo_city_state,
    on="geolocation_zip_code_prefix",
    how="left")

geo_clean.head()
```

```python
print("Rows:", len(geo_clean))
print("Unique ZIP prefixes:",geo_clean["geolocation_zip_code_prefix"].nunique())
```

```python
# geolocation coverage for customers

customer_geo_coverage = customers["customer_zip_code_prefix"].isin(geo_clean["geolocation_zip_code_prefix"])

print("Customers with geolocation match:",customer_geo_coverage.sum())

print("Customers without geolocation match:",(~customer_geo_coverage).sum())
```

```python
# geolocation coverage for sellers

seller_geo_coverage = sellers["seller_zip_code_prefix"].isin(geo_clean["geolocation_zip_code_prefix"])

print("Sellers with geolocation match:",seller_geo_coverage.sum())

print("Sellers without geolocation match:",(~seller_geo_coverage).sum())
```
